```python
df = spark.read.format("delta").load(delta_path)

silver_df = (df.filter(
    (df["transactionId"].isNotNull()) &
    (df["userId"].isNotNull()) &
    (df["cardNumber"].isNotNull()) &
    (df["amount"].isNotNull()) &
    (df["location"].isNotNull()) &
    (df["date"].isNotNull()) &
    (df["time"].isNotNull()) &
    (df["cardType"].isNotNull()) &
    (df["merchant"].isNotNull()) &
    (df["payment_status"].isNotNull()) &
    (df["is_fraud"].isNotNull())
).dropDuplicates(["transactionId"])
)
```

```python
from pyspark.sql.functions import *

silver_mask_df = silver_df.withColumn(
    "cardNumber",
    concat(lit("************"), substring(col("cardNumber"), -4, 4))
)

delta_path = f"wasbs://{container}@{storage_account}.blob.core.windows.net/silver_pos_transactions/"

silver_mask_df.write.format("delta").option("mergeSchema", "true").mode("overwrite").save(delta_path)

display(silver_mask_df)
```

```python
#Creating a gold tables

# 1.user summary
df = spark.read.format("delta").load(f"wasbs://{container}@{storage_account}.blob.core.windows.net/silver_pos_transactions/")
# display(df)
delta_path = f"wasbs://{container}@{storage_account}.blob.core.windows.net/gold_pos_transactions/"

from pyspark.sql.functions import *
from pyspark.sql.types import *

user_summary_df = silver_df.groupBy("userId").agg(
    count("transactionId").alias("total_transactions"),
    sum("amount").alias("total_amount"),
    avg("amount").alias("avg_amount"))

user_summary_df.write.format("delta") \
    .mode("overwrite") \
    .save(delta_path + "user_summary_df.delta/")
```

```python
fraud_status = df.groupBy("date").agg(
    count("*").alias("total_txns"),
    sum(when(col("is_fraud") == "yes", 1).otherwise(0)).alias("fraud_txns")
).withColumn("fraud_rate", col("fraud_txns") / col("total_txns"))
display(fraud_status)
fraud_status.write.format("delta") \
    .mode("overwrite") \
    .save(delta_path + "fraud_stats.delta/")
```

```python
location_summary = df.groupBy("location").agg(
    count("*").alias("total_txns"),
    sum(when(col("is_fraud") == "yes", 1).otherwise(0)).alias("fraud_txns")
).withColumn("fraud_rate", col("fraud_txns") / col("total_txns"))
display(location_summary)
location_summary.write.format("delta") \
    .mode("overwrite") \
    .save(delta_path + "location_summary.delta/")
```

```python
payment_status = df.groupBy("payment_status").agg(
    count("*").alias("total_txns"),
    sum(when(col("is_fraud") == "yes", 1).otherwise(0)).alias("fraud_txns")
)
display(payment_status)
payment_status.write.format("delta") \
    .mode("overwrite") \
    .save(delta_path + "payment_status.delta/")
```
